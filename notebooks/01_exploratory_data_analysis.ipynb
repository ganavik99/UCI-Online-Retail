```python
import pandas as pd

df = pd.read_excel("../data/raw/online_retail.xlsx")
```

```python
data = df.copy()
```

```python
data["IsReturn"] = data["InvoiceNo"].astype(str).str.startswith("C")
```

```python
data["IsReturn"].value_counts()
```

```python
returns = data[data["IsReturn"] == True].copy()
```

```python
returns.head()
```

```python
returns.shape
```

```python
sales = data[data["IsReturn"] == False].copy()
```

```python
sales.head()
```

```python
sales.shape
```

```python
sales = sales.drop_duplicates().copy()
returns = returns.drop_duplicates().copy()
```

```python
print("Sales duplicates:", sales.duplicated().sum())
print("Returns duplicates:", returns.duplicated().sum())
```

```python
sales = sales[sales["UnitPrice"] > 0].copy()
```

```python
sales["UnitPrice"].describe()
```

```python
sales = sales[sales["Quantity"] > 0].copy()
```

```python
sales["Quantity"].describe()
```

```python
returns = returns[returns["UnitPrice"] > 0].copy()
```

```python
sales["Revenue"] = sales["Quantity"] * sales["UnitPrice"]
```

```python
returns["ReturnValue"] = (
    returns["Quantity"].abs() * returns["UnitPrice"]
)
```

```python
sales["InvoiceDate"] = pd.to_datetime(sales["InvoiceDate"])

sales["Year"] = sales["InvoiceDate"].dt.year
sales["Month"] = sales["InvoiceDate"].dt.month
sales["MonthName"] = sales["InvoiceDate"].dt.month_name()
sales["DayOfWeek"] = sales["InvoiceDate"].dt.day_name()
```

```python
returns["InvoiceDate"] = pd.to_datetime(returns["InvoiceDate"])

returns["Year"] = returns["InvoiceDate"].dt.year
returns["Month"] = returns["InvoiceDate"].dt.month
returns["MonthName"] = returns["InvoiceDate"].dt.month_name()
returns["DayOfWeek"] = returns["InvoiceDate"].dt.day_name()
```

```python
sales = sales.dropna(subset=["Description"]).copy()
returns = returns.dropna(subset=["Description"]).copy()
```

```python
customer_sales = sales.dropna(subset=["CustomerID"]).copy()
```

```python
print("SALES DATASET")
print("=" * 40)
print("Rows:", len(sales))
print("Columns:", len(sales.columns))
print("Revenue:", sales["Revenue"].sum())

print("\nRETURNS DATASET")
print("=" * 40)
print("Rows:", len(returns))
print("Columns:", len(returns.columns))
print("Return value:", returns["ReturnValue"].sum())
```

```python
print("Sales negative quantities:", (sales["Quantity"] < 0).sum())
print("Sales invalid prices:", (sales["UnitPrice"] <= 0).sum())
print("Sales duplicates:", sales.duplicated().sum())

print("Returns negative quantities:", (returns["Quantity"] < 0).sum())
print("Returns invalid prices:", (returns["UnitPrice"] <= 0).sum())
print("Returns duplicates:", returns.duplicated().sum())
```

```python
sales.to_csv(
    "../data/processed/clean_sales.csv",
    index=False
)

returns.to_csv(
    "../data/processed/returns.csv",
    index=False
)
```
